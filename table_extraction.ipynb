Read a .csv from a URL with Pandas

Target website: https://www.football-data.co.uk/data.php

```python
import pandas as pd
```

```python
# reading 1 csv file from the website
df_premier26 = pd.read_csv('https://football-data.co.uk/mmz4281/2627/E0.csv')
```

```python
# showing dataframe
df_premier26
```

```python
# rename columns
df_premier26 = df_premier26.rename(columns={'Date':'date',
                                            'HomeTeam':'home_team',
                                            'AwayTeam':'away_team',
                                            'FTHG': 'home_goals',
                                            'FTAG': 'away_goals'})
```
